# Genskab Hubbles figur (Øvelse 1)

Denne notebook laver en lineær regression mellem afstand og hastighed for galakser  
og finder Hubble-konstanten, baseret på datafiler som fx `hubble_tab1.dat`.  

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Vælg fil med galaksedata
inputfil = "hubble_tab1.dat"                  # <-- ændr her for at bruge andre filer
# inputfil = "hubble_tab1_solkorr.dat"
# inputfil = "hubble_tab1_modern.dat"
# inputfil = "hubble_tab1_modern_solkorr.dat"
# inputfil = "cosmicflows.dat"
# inputfil = "cosmicflows_adj.dat"
# inputfil = "pantheon_sh0es.dat"
```

```python
# Indlæs data
dist, vel = np.loadtxt(inputfil, unpack=True)

# Definér lineær funktion til regression
def linear(x, a):
    return a * x

# Fit funktion til data
popt, pcov = curve_fit(linear, dist, vel)
H0 = popt[0]
sigH0 = np.sqrt(pcov[0][0])

print(f"Bedste fit for Hubble-konstanten er H0 = {H0:.1f} ± {sigH0:.1f} km/s/Mpc")
```

```python
d_akse = np.linspace(0, 1.1 * max(dist))
bestfit = linear(d_akse, H0)

plt.clf()
if len(vel) < 100:
    plt.scatter(dist, vel, color='b')
else:
    plt.scatter(dist, vel, color='b', s=1, alpha=.1)

plt.plot(d_akse, bestfit, 'r')
plt.xlabel('Afstand / Mpc')
plt.ylabel('Hastighed / km/s')
plt.title(inputfil)
plt.show()
```
